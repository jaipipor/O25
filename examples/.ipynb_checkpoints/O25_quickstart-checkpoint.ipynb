# O25 quick start
Correct one 12-band OLCI-like spectrum and inspect BH4 uncertainty.

```python
import numpy as np
import o25
from pathlib import Path
```

```python
data = np.loadtxt(Path('data/example_spectrum.csv'), delimiter=',', skiprows=1)
wl, Rrs = data[:,0], data[:,1]
r = o25.correct(wl, Rrs, g0=(30,20,120))
r.uncertainty_status, r.k
```

`r.u` is defined at the 12 matched BH4 bands. For extra hyperspectral wavelengths it is `NaN`, because BH4 uncertainty is not spectrally interpolated.
